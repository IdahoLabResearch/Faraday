```python
from warehouse.models import Node, Class, Ontology 
```

```python
# Assuming you have a Class instance and an Ontology instance
class_instance = Class.objects.get(id=1)  # Replace 1 with the actual ID
ontology_instance = Ontology.objects.get(id=1)  # Replace 1 with the actual ID

# Query to retrieve all nodes with the specified class and ontology
nodes = Node.objects.filter(cls=class_instance, ontology=ontology_instance)

# Print the results
print(nodes)
```
